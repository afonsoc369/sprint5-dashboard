# Análise exploratória — anúncios de veículos\n\nEste notebook faz uma exploração inicial do ficheiro `vehicles_us.csv` com Pandas e Plotly Express.\n\n> Antes da entrega, confirme que está a utilizar o CSV oficial fornecido pela TripleTen. O CSV incluído neste pacote é demonstrativo.

```python
import pandas as pd\nimport plotly.express as px\n\nvehicles = pd.read_csv('../vehicles_us.csv')\nvehicles.head()
```

## Estrutura e qualidade dos dados

```python
vehicles.info()\ndisplay(vehicles.describe(include='all').T)\ndisplay(vehicles.isna().sum().sort_values(ascending=False))
```

## Distribuição dos preços

```python
fig = px.histogram(vehicles, x='price', nbins=40, title='Distribuição dos preços dos veículos')\nfig.show()
```

## Relação entre preço e quilometragem

```python
fig = px.scatter(vehicles, x='odometer', y='price', color='type',\n                 title='Preço vs. quilometragem',\n                 hover_data=['model', 'model_year', 'condition'])\nfig.show()
```

## Preço por tipo de veículo

```python
fig = px.box(vehicles, x='type', y='price', color='type',\n             title='Distribuição dos preços por tipo de veículo')\nfig.show()
```
